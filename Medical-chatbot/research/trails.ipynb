```python
def split_text(text, chunk_size=500, overlap=50):
    chunks = []
    start = 0
    while start < len(text):
        end = start + chunk_size
        chunks.append(text[start:end])
        start += chunk_size - overlap
    return chunks

all_text = "Your large medical text here..."  # can combine multiple docs
text_chunks = split_text(all_text)

vectors = []

for i, doc in enumerate(text_chunks):
    vectors.append((
        f"medicalbot-{i}",
        embed_model.encode(doc).tolist(),
        {"text": doc}
    ))

index.upsert(vectors)
print("✅ Upsert done:", len(vectors))
```

```python
import random
from pinecone import Pinecone
from sentence_transformers import SentenceTransformer

# 1. Setup
PINECONE_API_KEY = "pcsk_4PV3VT_Tq2YvVTp6u4LTqMX8NgwmniTExRdw4RfHSi2S5qBhFmDAuXu2Pm93uB4yeKHTDU"
pc = Pinecone(api_key=PINECONE_API_KEY)
index = pc.Index(name="medicalbot", host="medicalbot-a9dxuoz.svc.aped-4627-b74a.pinecone.io")
model = SentenceTransformer('all-MiniLM-L6-v2')

# 2. Clear old data (Very important to avoid mixing!)
index.delete(delete_all=True)
print("🗑️ Index cleared.")

# 3. Create 1,000 TOTALLY UNIQUE records
data_to_upsert = []
symptoms = ["chronic inflammation", "acute rash", "persistent cough", "severe fatigue", "joint stiffness"]
body_parts = ["knees", "shoulders", "lower back", "chest", "abdominal wall"]

for i in range(1, 1001):
    # We combine the index 'i' into the text so it is 100% unique
    symptom = random.choice(symptoms)
    location = random.choice(body_parts)
    unique_text = f"Patient Case {i}: Exhibit shows {symptom} in the {location}. Refer to Medical_book.pdf section {i}."
    
    data_to_upsert.append({
        "id": f"vec-{i}",
        "values": [], # Will be filled by model
        "metadata": {
            "page": i % 500,  # Cycles through pages 0-499
            "source": "Data\\Medical_book.pdf",
            "text": unique_text
        }
    })

# 4. Batch Upsert
batch_size = 100
for i in range(0, len(data_to_upsert), batch_size):
    batch = data_to_upsert[i : i + batch_size]
    # Extract only the text for embedding
    texts = [item["metadata"]["text"] for item in batch]
    embeddings = model.encode(texts).tolist()
    
    # Put the embeddings back into the batch
    for j in range(len(batch)):
        batch[j]["values"] = embeddings[j]
    
    index.upsert(vectors=batch)
    print(f"✅ Uploaded batch {i//batch_size + 1} (Records {i} to {i+100})")

print("\n✨ Done! Now check your Pinecone Browser. Every row will be unique.")
```

```python
import os
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_pinecone import PineconeVectorStore

file_path = r"C:\Users\tharu\OneDrive\Desktop\end to end chatbot\Medical-chatbot\Data\medicine.pdf"

if not os.path.exists(file_path):
    print(f"❌ Still can't find it! Path tried: {file_path}")
    print("Check if 'medicine.pdf' is actually inside that 'Data' folder.")
else:
    print("✅ File found! Starting the upload...")
    try:

        loader = PyPDFLoader(file_path)
        data = loader.load()
        
        text_splitter = RecursiveCharacterTextSplitter(chunk_size=700, chunk_overlap=70)
        docs = text_splitter.split_documents(data)
        
        embeddings = HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")
        print(f"🚀 Uploading {len(docs)} chunks from the book...")
        vectorstore = PineconeVectorStore.from_documents(
            docs, 
            embeddings, 
            index_name="medicalbot",
            text_key="text"
        )
        print("✅ DONE! Your 'medicine.pdf' is now inside the MedicalBot.")

    except Exception as e:
        print(f"❌ Error during upload: {e}")
```
